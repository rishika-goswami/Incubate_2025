```python
import re
from sklearn.preprocessing import MultiLabelBinarizer

def split_gram(x):
    if pd.isnull(x):
        return []
    
    return [v.strip().lower() for v in x.split(',') if v.strip()]


fdf['Gram_List'] = fdf['Gram Stain'].apply(split_gram)


mlb_gram = MultiLabelBinarizer()
gram_dummies = pd.DataFrame(
    mlb_gram.fit_transform(fdf['Gram_List']),
    columns=[f"Gram_{c}" for c in mlb_gram.classes_],
    index=fdf.index
)

fdf = pd.concat([fdf, gram_dummies], axis=1)



def clean_org(x):
    if pd.isnull(x):
        return ''
    s = x.lower()
    
    s = re.sub(r'\([^)]*\)', '', s)
    
    s = re.sub(r'[^a-z0-9,; ]+', ' ', s)
    return s

def split_org(s):
    parts = re.split(r'[;,]| and |\d+\)', s)
    return [p.strip() for p in parts if p.strip()]


fdf['Org_Clean'] = fdf['Organism Isolated'].apply(clean_org)
fdf['Org_List']  = fdf['Org_Clean'].apply(split_org)


mlb_org = MultiLabelBinarizer()
org_dummies = pd.DataFrame(
    mlb_org.fit_transform(fdf['Org_List']),
    columns=[f"Org_{c}" for c in mlb_org.classes_],
    index=fdf.index
)

fdf = pd.concat([fdf, org_dummies], axis=1)



def clean_specimen(x):
    if pd.isnull(x):
        return ''
    s = x.lower()
    
    s = re.split(r'\(|\d', s)[0]
    return s.strip()

fdf['Specimen_Clean'] = fdf['Specimen Type'].apply(clean_specimen)

# now it’s essentially a single‐label categorical variable
spec_dummies = pd.get_dummies(fdf['Specimen_Clean'], prefix='Spec')

fdf = pd.concat([fdf, spec_dummies], axis=1)





fdf.drop(columns=['Gram_List', 'Org_Clean', 'Org_List', 'Specimen_Clean'], inplace=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report


antibiotics = [
    "Amikacin", "Amoxicillin", "Amoxicillin+Clavulanic Acid", "Ampicillin",
    "Ampicillin+Sulbactam", "Azithromycin", "Aztreonam", "Benzyl Penicillin",
    "Cefazolin", "Cefepime", "Cefotaxime", "Cefoxitin", "Ceftazidime",
    "Ceftriaxone", "Cefuroxime", "Cephalexin", "Chloramphenicol",
    "Ciprofloxacin", "Clarithromycin", "Clindamycin", "Colistin",
    "Co-Trimoxazole", "Doripenem", "Ertapenem", "Erythromycin", "Gentamicin",
    "Imipenem", "Levofloxacin", "Linezolid", "Meropenem", "Norfloxacin",
    "Nitrofurantoin", "Piperacillin- Tazobactam", "Tigecycline",
    "Tobramycin", "Trimethoprim", "Trimethoprim/Sulfame Thoxale", "Vancomycin"
]

#  filling missing antibiotic results with "dc" 
fdf[antibiotics] = fdf[antibiotics].fillna("dc")

# separating features (X) and targets (y) 
X = fdf.drop(columns=antibiotics)
y = fdf[antibiotics].copy()

# one‐hot‐encoding all remaining categorical features 
X_enc = pd.get_dummies(X, drop_first=True)

# label‐encoding each antibiotic target to numeric classes 
label_encoders = {}
y_enc = pd.DataFrame(index=y.index)
for col in antibiotics:
    le = LabelEncoder()
    # fit on strings like 'R', 'S', 'dc', etc.
    y_enc[col] = le.fit_transform(y[col])
    label_encoders[col] = le

# train/test split 
X_train, X_test, y_train, y_test = train_test_split(
    X_enc, y_enc, test_size=0.2, random_state=42
)

# build & train the multi‐output Random Forest 
base_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)
multi_rf = MultiOutputClassifier(base_rf, n_jobs=-1)
multi_rf.fit(X_train, y_train)

─
y_pred = multi_rf.predict(X_test)

accuracies = {}
for idx, col in enumerate(antibiotics):
    acc = accuracy_score(y_test[col], y_pred[:, idx])
    accuracies[col] = acc
    print(f"{col:30s} accuracy: {acc:.3f}")


avg_acc = sum(accuracies.values()) / len(accuracies)
print(f"\nAverage accuracy across all antibiotics: {avg_acc:.3f}")



print("\nClassification Reports:")
for idx, col in enumerate(antibiotics):
    print(f"\n--- {col} ---")
    # y_test[col] is your true (encoded) labels, y_pred[:, idx] your predictions
    print(classification_report(
        y_test[col],
        y_pred[:, idx]
    ))
```

```python
import pandas as pd
import numpy as np
import pprint

# 1.  reconstructing the true feature list straight from the model


first_tree = multi_rf.estimators_[0]

if hasattr(first_tree, "feature_names_in_"):
    train_cols = list(first_tree.feature_names_in_)
else:                                   # fallback (older scikit-learn)
    # Here we *must* already have saved train_cols earlier.  If not,
    # raise an explicit error so you know to set it.
    try:
        train_cols = train_cols         # noqa: F821
    except NameError:
        raise RuntimeError(
            "train_cols not found — save it right after you build X_enc:\n"
            "   train_cols = list(X_enc.columns)"
        )

expected_len = len(train_cols)
print(f"Model expects {expected_len} features.")


# 2.  building a zero-filled Series with ALL expected columns

def zero_feature_vector():
    return pd.Series(0, index=train_cols, dtype=float)


# 3.  helper for turning a raw 9-field dict into that vector

def feature_vector_from_raw(raw):
    fv = zero_feature_vector()

    # numeric survivors
    for num in ["UMR No.", "Age", "Colony Count"]:
        if num in fv.index:
            fv[num] = float(raw[num])

    # simple categoricals
    def activate(base, val):
        col = f"{base}_{val.strip()}"
        if col in fv.index:
            fv[col] = 1
    activate("Reports Photo", raw["Reports Photo"])
    activate("Gender",        raw["Gender"])
    activate("Department",    raw["Department"])

    # Gram Stain (multi)
    for g in split_gram(raw["Gram Stain"]):
        col = f"Gram_{g}"
        if col in fv.index:
            fv[col] = 1

    # Organism Isolated (multi)
    for o in split_org(clean_org(raw["Organism Isolated"])):
        col = f"Org_{o}"
        if col in fv.index:
            fv[col] = 1

    # Specimen Type (single)
    spec = clean_specimen(raw["Specimen Type"])
    spec_col = f"Spec_{spec}"
    if spec_col in fv.index:
        fv[spec_col] = 1

    return fv.to_numpy().reshape(1, -1)   # shape (1, expected_len)


# 4.  prediction wrapper

def predict_antibiotics(raw):
    X_vec = feature_vector_from_raw(raw)
    # final safety check
    assert X_vec.shape[1] == expected_len, \
        f"Vector length {X_vec.shape[1]} vs model expects {expected_len}"
    codes = multi_rf.predict(X_vec)[0]
    return {
        ab: label_encoders[ab].inverse_transform([codes[i]])[0]
        for i, ab in enumerate(antibiotics)
    }


# 5.  testing with manual samples

'''sample = {
    "UMR No.":             695323,
    "Reports Photo":      "no sensitivity data",
    "Age":                   16,
    "Gender":              "male",
    "Gram Stain":          "no organisms seen",
    "Organism Isolated":   "no growth",
    "Specimen Type":       "blood bag unit no 2788 plt (00c13691",
    "Department":          "unknown",
    "Colony Count":         0
}'''
sample = {
    "UMR No.":             646255,
    "Reports Photo":      "yes",
    "Age":                   39,
    "Gender":              "female",
    "Gram Stain":          "no pus cells, gram negative bacilli seen",
    "Organism Isolated":   "acinetobacter spp",
    "Specimen Type":       "urine",
    "Department":          "general medicine",
    "Colony Count":        "unknown"
}


pprint.pprint(predict_antibiotics(sample))
```
